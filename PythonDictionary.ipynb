<h2>Python Variables:</h2>
<p>These are variables types which are used in python. Most languages have all of these variables types as well.</p>
<ul>
    <li>Integer: Basic number with no decimal places</li>
    <li>Float: Basic number with decimal places (i.e. 12.97856)</li>
    <li>Double: Basic number with decimal places. Can contain double the decimal places of a float</li>
    <li>String: A string of characters (i.e. "Hello world!")</li>
    <li>Boolean: A variable that is either true or false</li>
    <li>Array: A list of numbers which can be accessed via element (i.e. array defined: grades = [90, 85, 100, 76], array accessed: grades[2] returns 100)</li>
    <li>List: A list of anything which can be accessed via element</li>
</ul>
