<h1><u>Python Dictionary</u></h1>
<h2>Python Variables:</h2>
<p>These are variables types which are used in python. Most languages have all of these variables types as well.</p>
<ul>

    <li>Integer: Basic number with no decimal places</li>
    <li>Float: Basic number with decimal places (i.e. 12.97856)</li>
    <li>Double: Basic number with decimal places. Can contain double the decimal places of a float</li>
    <li>String: A string of characters (i.e. "Hello world!")</li>
    <li>Boolean: A variable that is either true or false</li>
    <li>Array: A list of numbers which can be accessed via element (i.e. array defined: grades = [90, 85, 100, 76], array accessed: grades[2] returns 100)</li>
    <li>List: A list of anything which can be accessed via element</li>
</ul>

<h2>Basic Statements</h2>

```python
if(1==1): #Basic if statement. Evaluates statement inside
    print("goodbye world")
elif(2==2): #Basic else if statement. Evaluates the statement inside only if the previous if fails
    print("I will never be reached :cry:")
else: #Basic else statement. Only excecutes if both above statements fail
    print("you have a better chance than me")
```

```python
while(1==1): #Executes code while the statement inside is true. Will execute forever
    print("yay")
```

```python
for i in range(10): #Executes code the number of times specified
    print("there will be 10 of these")
```

<h2>Basic Comparison Operators</h2>
<p>Operations we can use in an if, elif, or while statement</p>
<ul>
    <li>==: Equal. Evaluates to true if they are equal</li>
    <li>!=: Not Equal. Evaluates to true if they are not equal</li>
    <li>< : Less than. Evaluates to true if the left number is less than the right number</li>
    <li><= : Less than or equal true. Evaluate to true is the left number is less than or equal to the right number</li>
    <li>> : More than. Evaluates to true if the left number is more than the right number</li>
    <li>>= : More than or equal to. Evaluates to true if the left number is more than or equal to the right number</li>
    <li>not : Added on to a boolean statement to reverse the evaluation (i.e. not (1==1))</li>
    <li>and : Used to append 2 boolean statements (i.e. (1==1 and 2==2))</li>
</ul>

<h2>Basic Operations</h2>
<p>Operations we can use to change variables</p>
<ul>
    <li>+</li>
    <li>-</li>
    <li>+=: Makes a variable equal to its current value plus a number (i.e. if x = 1, x += 2 makes x = 3)</li>
    <li>-=: Makes a variable equal to its current value minus a number</li>
    <li>*: Multiplication</li>
    <li>/: Division</li>
    <li>*=: makes a variable equal to its current value times a number</li>
    <li>/=: makes a variable equal to its current value divided by a number</li>
</ul>
